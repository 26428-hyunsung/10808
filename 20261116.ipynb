```python
def a(num):
 if num % 2 == 0:
  return '짝수'
 if num % 2 == 1:
  return '홀수'

b = int(input())
print(a(b))
```

```python
def t(num,num2):
  t = 0.5 * num * num2
  return t
b = int(input('삼각형의 밑변을 입력하세요:'))
h = int(input('삼각형의 높이를 입력하세요:'))
result = t(b,h)
print(f'삼각형의 넓이는 {result}입니다')
```

```python
concat = lambda s1,s2: s1 + '' + s2
result = concat('파이썬','프로그래밍')
print('문자열 합치기:',result)
```

```python
greeting = lambda name, time: f'안녕하세요,{name}님 좋은 {time}입니다'
result = greeting('김현성','아침')
print(result)
```
