```python
list1 = ['우유','빵','버터','빵','치즈','과자','우우']
list2 = ['빵','과일','야채','버터','우유','커피']
print('리스트1 항목(중복 제거):', set(list1))
print('리스트1,2 공통 항목:',set(list1)&set(list2))
```

```python
sentence1 = input('첫번째 영어 문장을 입력하세요:')
sentence2 = input('두번째 영어 문장을 입력하세요:')
sentence3 = input('세번째 영어 문장을 입력하세요:')
text = sentence1 + '' + sentence2 + '' + sentence3
text = text.replace(',','')
text = text.replace('.','')
text = text.lower()
words = text.split()
unique_words = set(words)
sorted_unique_words = sorted(unique_words)
print('\n단어목록')
print(sorted_unique_words)
```

```python
import random
mem = ['김민준','이서준','박도윤','최시우','정예준','강주원','조유준']
winers_cnt = int(input('뽑을 winers 입력하세요:'))
winers = random.random.sample(mem.winers_cnt)
winers.winers.sort()
print(winers)
```
